# Experiment No.3: Pre-processing (Normalization & Encoding)

### **Objective:**
To pre-process a dataset, including normalization and encoding, for neural network input.

### **Concept:**
Neural Networks are highly sensitive to the scale of features because they use Gradient Descent.
1. **Standardization :** Scales features to have $\mu=0$ and $\sigma=1$.
2. **One-Hot Encoding :** Converts categorical strings into numerical vectors. This prevents the model from assuming an artificial order (e.g., 'Paris' > 'London').

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
#LOAD THE REAL WORLSD DATASETS( Tips dataset)
df_tips = sns.load_dataset('tips')
print("Original Dataset Preview (First 5 rows) : ")
display(df_tips.head())
```

```python
from pandas.core.arrays.categorical import Categorical
#define the trnsformations
#we scale numerical values Total Bill,size
#we encode categorical values: sex, smoker, day , time

numeric_features =['total_bill','size']
categorical_features = ['sex','smoker','day','time']


ct = ColumnTransformer([
    ('scaler',StandardScaler(),numeric_features),
    ('encoder',OneHotEncoder(),categorical_features)


    ])
```

```python
#3.Apply Transformations

df_tips = sns.load_dataset('tips')
processed_data = ct.fit_transform(df_tips)
```
